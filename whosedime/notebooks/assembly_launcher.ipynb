# 🎬 Whose Dime — assembly (launcher)

Assembles a video from its Drive folder into `render/`.

1. Fill in the form: first `MODE = "extract"` with `SCENES = "1-10"`, then `MODE = "full"`.
2. **Runtime → Run all**, and allow Google Drive. No GPU needed.

This file is a **launcher**: it runs the real notebook, `whosedime/notebooks/assembly_whosedime.ipynb`, from the `main` branch of the repo (decision D5), and prints the commit it runs. It never needs copying again.

```python
# @title 🎬 Whose Dime — assembly { display-mode: "form" }
VIDEO = ""  # @param {type:"string"}
# Empty = the newest video with a voice and no finished render. Else the folder name: "01_uber".

MODE = "extract"  # @param ["plan", "extract", "full"]
# plan: check and time everything, no render · extract: the scenes below only · full: the whole video

SCENES = "1-10"  # @param {type:"string"}
# With MODE = "extract": first-last scene numbers.

ROOT = "/content/drive/MyDrive/WhoseDime"  # @param {type:"string"}
# The WhoseDime folder in your Drive.

# ── Do not change below ──
BRANCH = "main"  # decision D5: Colab always runs main
import subprocess, sys
subprocess.run(["rm", "-rf", "/content/Channel"], check=True)
if subprocess.run(["git", "clone", "-q", "--depth", "1", "-b", BRANCH,
                   "https://github.com/Johnfrancis6/Channel.git", "/content/Channel"]).returncode:
    raise SystemExit(f"❌ Could not fetch the code (branch {BRANCH}).")
COMMIT = subprocess.run(["git", "-C", "/content/Channel", "log", "-1", "--format=%h %ad %s", "--date=short"],
                        capture_output=True, text=True).stdout.strip()
print(f"▶ Running {BRANCH} @ {COMMIT}")
sys.path.insert(0, "/content/Channel/whosedime/notebooks")
sys.modules.pop("launcher", None)
import launcher
launcher.run("assembly_whosedime.ipynb", dict(VIDEO=VIDEO, MODE=MODE, SCENES=SCENES, ROOT=ROOT), globals())
```
