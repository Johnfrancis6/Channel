# 🖼️ Whose Dime — images and clips (launcher)

Generates the scene images (Gemini API, Nano Banana 2.1, 2K) and the clips (Gemini Omni Flash, 1080p, 8 s) of a video **straight into its Drive folder** (`images/`, `clips/`), from the prompts of `03_scenes.md`. No GPU needed.

1. Once: in Colab, **🔑 Secrets** (left bar) → add `GEMINI_API_KEY` (a key from a project with billing; these models have no free tier) and allow this notebook to read it.
2. Fill in the form: first `MODE = "estimate"` (free), then `"test"` (3 images, 1 clip, about $1), then `"full"` with the amount you agree in `APPROVED_USD`.
3. **Runtime → Run all**, and allow Google Drive.

This file is a **launcher**: it runs `whosedime/notebooks/generate_whosedime.ipynb` from `main` (decision D5) and prints the commit it runs. Every paid call is a line of `generate/ledger.jsonl` in the video folder; `generate/status.json` says what is done and what is next.

```python
# @title 🖼️ Whose Dime — images and clips { display-mode: "form" }
VIDEO = ""  # @param {type:"string"}
# The video folder name, e.g. "01_uber" (it must hold 03_scenes.md).

MODE = "estimate"  # @param ["estimate", "test", "full"]
# estimate: the price of what is missing, no call · test: 3 images and 1 clip at most · full: every missing image and clip

APPROVED_USD = 0.0  # @param {type:"number"}
# With MODE = "full": the amount you agree to spend on this video (decision D18). The run stops before passing it.

ROOT = "/content/drive/MyDrive/WhoseDime"  # @param {type:"string"}
# The WhoseDime folder in your Drive.

# ── Do not change below ──
BRANCH = "main"  # decision D5: Colab always runs main
import subprocess, sys
subprocess.run(["rm", "-rf", "/content/Channel"], check=True)
if subprocess.run(["git", "clone", "-q", "--depth", "1", "-b", BRANCH,
                   "https://github.com/Johnfrancis6/Channel.git", "/content/Channel"]).returncode:
    raise SystemExit(f"❌ Could not fetch the code (branch {BRANCH}).")
COMMIT = subprocess.run(["git", "-C", "/content/Channel", "log", "-1", "--format=%h %ad %s", "--date=short"],
                        capture_output=True, text=True).stdout.strip()
print(f"▶ Running {BRANCH} @ {COMMIT}")
sys.path.insert(0, "/content/Channel/whosedime/notebooks")
sys.modules.pop("launcher", None)
import launcher
launcher.run("generate_whosedime.ipynb", dict(VIDEO=VIDEO, MODE=MODE, APPROVED_USD=APPROVED_USD, ROOT=ROOT), globals())
```
