# 🎙️ Whose Dime — voice (launcher)

Makes the voice of a video from `02_script_voice.txt`, into `voice/`.

1. Fill in the form (usually: only `MODE`).
2. **Runtime → Run all**, and allow Google Drive. A **GPU** is needed: Runtime → Change runtime type → T4 GPU.

This file is a **launcher**: it runs the real notebook, `whosedime/notebooks/voice_whosedime.ipynb`, from the `main` branch of the repo (decision D5), and prints the commit it runs. It never needs copying again.

Order: `reference` (first time: your ElevenLabs clip in `Voice/elevenlabs/`, D22; `design` is the old fallback), then for each video `sample` (listen), then `full` with `I_LISTENED` ticked. `blind_test` on video 1 only.

```python
# @title 🎙️ Whose Dime — voice { display-mode: "form" }
VIDEO = ""  # @param {type:"string"}
# Empty = the newest video with a 02_script_voice.txt and no ready voice. Else the folder name: "01_uber".

MODE = "sample"  # @param ["reference", "design", "sample", "full", "blind_test"]
# reference: the narrator from your ElevenLabs clip in Voice/elevenlabs/ (D22) · design: the old VoiceDesign narrator · sample: the 20 s sample · full: the whole voice · blind_test: video 1 only

I_LISTENED = False  # @param {type:"boolean"}
# Tick after listening to voice/listen_20s.wav. "full" refuses to start without it (D6).

KEEP_VOICE = ""  # @param ["", "A", "B", "C"]
# With MODE = "design": the candidate to keep as the narrator. Empty = make the candidates.

REDO_BLOCKS = ""  # @param {type:"string"}
# Block numbers to say again with another seed: "12, 40". Empty = none.

ROOT = "/content/drive/MyDrive/WhoseDime"  # @param {type:"string"}
# The WhoseDime folder in your Drive.

# ── Do not change below ──
BRANCH = "main"  # decision D5: Colab always runs main
import subprocess, sys
subprocess.run(["rm", "-rf", "/content/Channel"], check=True)
if subprocess.run(["git", "clone", "-q", "--depth", "1", "-b", BRANCH,
                   "https://github.com/Johnfrancis6/Channel.git", "/content/Channel"]).returncode:
    raise SystemExit(f"❌ Could not fetch the code (branch {BRANCH}).")
COMMIT = subprocess.run(["git", "-C", "/content/Channel", "log", "-1", "--format=%h %ad %s", "--date=short"],
                        capture_output=True, text=True).stdout.strip()
print(f"▶ Running {BRANCH} @ {COMMIT}")
sys.path.insert(0, "/content/Channel/whosedime/notebooks")
sys.modules.pop("launcher", None)
import launcher
launcher.run("voice_whosedime.ipynb", dict(VIDEO=VIDEO, MODE=MODE, I_LISTENED=I_LISTENED, KEEP_VOICE=KEEP_VOICE,
                                           REDO_BLOCKS=REDO_BLOCKS, ROOT=ROOT), globals())
```
